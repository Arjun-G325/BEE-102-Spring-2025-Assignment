```python
import pandas as pd
import matplotlib.pyplot as plt

col = ['chromosome', 'start', 'end', 'other1', 'other2', 'symbol']
df = pd.read_csv('query.bed.gz', sep='\t', compression='gzip',header=None,names=col, low_memory=False)

df['length']=df['end']-df['start']
len_count = df['length'].value_counts().sort_index()

norm_freq = len_count / len_count.sum()

plt.figure(figsize=(12, 6))
plt.plot(norm_freq.index, norm_freq.values, lw=1.5)
plt.xlabel("Fragment Length [bp]")
plt.ylabel("Normalized Frequency [A.U.]")
plt.title("Normalized Fragment Length Distribution")
plt.grid(True)
plt.tight_layout()
plt.show()
```
