```python
from collections import defaultdict
import pandas as pd

def build_markov_model_from_fasta(fasta_file):
    transitions = defaultdict(lambda: defaultdict(int))
    prev_base = None

    with open(fasta_file, "r") as file:
        for line in file:
            line = line.strip()
            if line.startswith(">"):
                continue
            for base in line.upper():
                if base not in "ACGT":
                    continue
                if prev_base:
                    transitions[prev_base][base] += 1
                prev_base = base

    nucleotides = ['A', 'C', 'G', 'T']
    matrix = pd.DataFrame(0.0, index=nucleotides, columns=nucleotides)

    for current in transitions:
        total = sum(transitions[current].values())
        for next_base in transitions[current]:
            matrix.at[current, next_base] = transitions[current][next_base] / total

    return matrix

fasta_path = "multiline_input.fasta"
transition_matrix = build_markov_model_from_fasta(fasta_path)

print("Transition Matrix:")
display(transition_matrix)
```
